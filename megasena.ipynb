```python

# Junta todas as dezenas em uma única série
dezenas = pd.concat([
    df['Bola1'],
    df['Bola2'],
    df['Bola3'],
    df['Bola4'],
    df['Bola5'],
    df['Bola6']
])

# Conta quantas vezes cada número apareceu
frequencia = dezenas.value_counts().sort_values(ascending=False)
#frequencia = dezenas.value_counts().sort_index()

# Transforma em DataFrame
frequencia_df = frequencia.reset_index()
frequencia_df.columns = ['Dezena', 'Quantidade']



```

```python
print (frequencia_df)
```

```python
# Salva em CSV
frequencia_df.to_csv(
    'frequencia_dezenas.csv',
    sep=';',
    index=False,
    encoding='utf-8-sig'
)

print("Arquivo gerado com sucesso!")
```

```python

# Pega as 60 dezenas mais frequentes
top60 = frequencia_df.head(60)

# Lista das dezenas
dezenas_top60 = top60['Dezena'].tolist()

jogos = []

for i in range(10):
    jogo = sorted(random.sample(dezenas_top60, 6))
    jogos.append(jogo)

# Exibe os jogos
for i, jogo in enumerate(jogos, start=1):
    print(f"Jogo {i}: {jogo}")
```

```python


top60 = frequencia_df.head(60)
dezenas_top60 = top60['Dezena'].tolist()

jogos = set()

while len(jogos) < 10:
    jogo = tuple(sorted(random.sample(dezenas_top60, 6)))
    jogos.add(jogo)

for i, jogo in enumerate(sorted(jogos), start=1):
    print(f"Jogo {i}: {list(jogo)}")
```

```python
df_jogos = pd.DataFrame(
    list(jogos),
    columns=['D1', 'D2', 'D3', 'D4', 'D5', 'D6']
)

df_jogos.to_csv(
    '10_jogos_megasena.csv',
    sep=';',
    index=False,
    encoding='utf-8-sig'
)

print ('Arquivo salvo com sucesso')
```

```python
# Cria a figura
fig, ax = plt.subplots(figsize=(12, 5))

# Remove os eixos
ax.axis('off')

# Cria a tabela
tabela = ax.table(
    cellText=df_jogos.values,
    colLabels=df_jogos.columns,
    loc='center'
)

# Ajusta tamanho da fonte
tabela.auto_set_font_size(False)
tabela.set_fontsize(10)

# Ajusta escala da tabela
tabela.scale(1.2, 1.5)

# Salva em PNG
plt.savefig(
    'jogos_megasena.png',
    bbox_inches='tight',
    dpi=300
)

plt.close()

print("Imagem salva como jogos_megasena.png")
```
