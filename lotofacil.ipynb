```python
dezenas = pd.concat([
    df['Bola1'],
    df['Bola2'],
    df['Bola3'],
    df['Bola4'],
    df['Bola5'],
    df['Bola6'],
    df['Bola7'],
    df['Bola8'],
    df['Bola9'],
    df['Bola10'],
    df['Bola11'],
    df['Bola12'],
    df['Bola13'],
    df['Bola14'],
    df['Bola15'],


])

frequencia = dezenas.value_counts().sort_values(ascending=False)

print(frequencia.head(30))
```

```python
frequencia = dezenas.value_counts()

mais_frequentes = frequencia[frequencia >= 2200]
medios = frequencia[(frequencia >= 1800) & (frequencia < 2199)]
menos_frequentes = frequencia[frequencia < 1800]

print ("Mais sorteados:")
print(mais_frequentes.head(50))
```

```python
# grafico

frequencia.sort_values(ascending=False).head(15).plot(kind='bar')

plt.title('15 Dezenas Mais Sorteadas')
plt.xlabel('Dezena')
plt.ylabel('Quantidade')
plt.show()
```

```python

frequencia = dezenas.value_counts()

mais_frequentes = frequencia[frequencia >= 2000]
medios = frequencia[(frequencia >= 1500) & (frequencia < 2000)]
menos_frequentes = frequencia[frequencia < 1500]

print("Mais sorteados:")
print(mais_frequentes.head(50))

# Pega apenas os 50 mais frequentes
top50 = frequencia.sort_values(ascending=False).head(50)
#top50 = mais_frequentes.head(50)

# Os índices são as dezenas
dezenas_top50 = top50.index.tolist()

jogos = []

for i in range(10):
    jogo = sorted(random.sample(dezenas_top50, 15))
    jogos.append(jogo)

for i, jogo in enumerate(jogos, start=1):
    print(f"Jogo {i}: {jogo}")
```

```python
df_jogos = pd.DataFrame(
    jogos,
    columns=[f'Dezena_{i}' for i in range(1, 16)]
)

df_jogos.to_csv(
    '10_jogos_lotofacil.csv',
    sep=';',
    index=False,
    encoding='utf-8-sig'
)

print ("Arquivo salvo com sucesso")
```
